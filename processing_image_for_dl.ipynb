NOW ,RESIZING THE IMAGE USING PILLOW LIBRARY

```python
from PIL import Image
img =Image.open('/content/trees.jpg?fit=fill&w=1200&h=630')
img_resize=img.resize((300,300))
img_resize.show()
```

```python
#displaying the image in numpy

img_res=mpimg.imread('/content/trees.jpg?fit=fill&w=1200&h=630')
img_res_plot=plt.imshow(img_res)
plt.show()
```

```python
#let's use opencv
import  cv2
img=cv2.imread('/content/trees.jpg?fit=fill&w=1200&h=630')
img_plot=plt.imshow(img)
plt.show()
```

```python
type(img)
```

```python
#let's convert it into grayscale image
img_gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
img_gray_plot=plt.imshow(img_gray,cmap='gray')
plt.show()
```

```python
print(img.shape)
print(img_gray.shape)
#it is clear now that the rgb factor has gone (that is the 3rd dimension)
```

```python
cv2.imshow(img_gray)
```
