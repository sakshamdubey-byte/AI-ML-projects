```python
# now we will split the data
from sklearn.model_selection import train_test_split

train_data,test_data=train_test_split(data,test_size=0.2,random_state=42)

# this is perfect
print(train_data.shape)
print(test_data.shape)
print(data.shape)
```

```python
!pip install transformers datasets seqeval -q
# intitally  without tokensier there was a one to one mapping of data labels-word
# but after introduincg toeknsier ,dfor example-bert ,it can split words ,leading to miscalcuation
# for example the situan becomes 10 labels and 15 words;for example



from transformers import AutoTokenizer
MODEL_NAME = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
def tokenize_and_align_labels(examples):
    tokenized_inputs = tokenizer(
        examples["tokens"],
        truncation=True,
        is_split_into_words=True,
        max_length=512
    )

    labels = []
    for i, label in enumerate(examples["labels"]):
        word_ids = tokenized_inputs.word_ids(batch_index=i)
        previous_word_idx = None
        label_ids = []
        for word_idx in word_ids:
            if word_idx is None:
                label_ids.append(-100)  # Special tokens like [CLS], [SEP]
            elif word_idx != previous_word_idx:
                label_ids.append(label_id[label[word_idx]])  # First subword gets real label ID
            else:
                label_ids.append(-100)  # Subsequent subwords ignored in loss calculation
            previous_word_idx = word_idx
        labels.append(label_ids)

    tokenized_inputs["labels"] = labels
    return tokenized_inputs
```

```python
# now  finally let's train and test data
from datasets import Dataset
# 1st we will convert pandas dataframes to hugging face dataset
train_h_f = Dataset.from_pandas(train_data)
test_h_f = Dataset.from_pandas(test_data)
# map tokenisation and alignment
train_tokenized=train_h_f.map(tokenize_and_align_labels,batched=True)
test_tokenized=test_h_f.map(tokenize_and_align_labels,batched=True)




```

```python
from transformers import AutoModelForTokenClassification, TrainingArguments, Trainer, DataCollatorForTokenClassification

# 1. Initialize pretrained model with your label mappings
model = AutoModelForTokenClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels=len(label_id),
    id2label=id_label,
    label2id=label_id
)

# 2. Dynamic padding collator
data_collator = DataCollatorForTokenClassification(tokenizer=tokenizer)

# 3. Training arguments
training_args = TrainingArguments(
    output_dir="./results",
    eval_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01,
    logging_steps=50,
    save_strategy="epoch"
)

# 4. Initialize Trainer (removed 'tokenizer' argument)
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tokenized,
    eval_dataset=test_tokenized,
    data_collator=data_collator,
)

# 5. Start training
trainer.train()
```

```python
import numpy as np
from seqeval.metrics import classification_report, f1_score

# Get predictions on validation set
predictions, labels, _ = trainer.predict(test_tokenized)
preds = np.argmax(predictions, axis=2)

# Filter out -100 (ignored tokens)
true_predictions = [
    [id_label[p] for (p, l) in zip(prediction, label) if l != -100]
    for prediction, label in zip(preds, labels)
]
true_labels = [
    [id_label[l] for (p, l) in zip(prediction, label) if l != -100]
    for prediction, label in zip(preds, labels)
]

print(classification_report(true_labels, true_predictions))
```

```python
import torch

def redact_pii(text):
    tokens = text.split()

    # 1. Tokenize pre-split words (returns a BatchEncoding object)
    inputs = tokenizer(tokens, is_split_into_words=True, return_tensors="pt", truncation=True, max_length=512)

    # 2. Extract word_ids BEFORE converting inputs to a dict
    word_ids = inputs.word_ids()

    # 3. Move tensors to GPU/device
    inputs = {k: v.to(model.device) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)

    predictions = torch.argmax(outputs.logits, dim=2)[0].cpu().numpy()

    redacted_tokens = []
    last_word_idx = None

    for idx, word_idx in enumerate(word_ids):
        # Skip [CLS], [SEP], or repeated subwords of the same token
        if word_idx is None or word_idx == last_word_idx:
            continue

        pred_label = id_label[predictions[idx]]

        if pred_label != 'O':
            redacted_tokens.append("[REDACTED]")
        else:
            redacted_tokens.append(tokens[word_idx])

        last_word_idx = word_idx

    return " ".join(redacted_tokens)

# Test on a sample string
sample_text = "Hello, my name is Sarah Conner and my email is sarah@example.com."
print("Original:", sample_text)
print("Redacted:", redact_pii(sample_text))
```
